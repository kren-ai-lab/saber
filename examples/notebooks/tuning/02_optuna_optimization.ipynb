# Hyperparameter Optimization with Optuna

This notebook demonstrates how to perform hyperparameter
optimization using the Optuna backend available in mlcore.

The examples cover:

- Binary classification
- Regression
- Custom search spaces
- Optimization result inspection

```python
from sklearn.datasets import (
    make_classification,
    make_regression,
)

from mlcore import MODEL_REGISTRY

from mlcore.core.search_space import (
    SearchSpace,
)

from mlcore.tuning.optuna import (
    OptunaOptimizer,
)
```

```python
X, y = make_classification(
    n_samples=1000,
    n_features=20,
    n_informative=10,
    n_redundant=5,
    random_state=42,
)
```

```python
spec = MODEL_REGISTRY.get(
    "random_forest"
)

spec.search_space
```

```python
optimizer = OptunaOptimizer(
    registry=MODEL_REGISTRY,
)
```

```python
result = optimizer.optimize(
    algorithm="random_forest",
    X=X,
    y=y,
    metric="accuracy",
    cv=5,
    n_trials=20,
    search_space=spec.search_space,
)
```

```python
print(
    f"Best score: {result.best_score:.4f}"
)

print(
    f"Best params: {result.best_params}"
)
```

```python
result.history[:5]
```

```python
study = result.study

study.best_trial
```

```python
X, y = make_regression(
    n_samples=1000,
    n_features=20,
    n_informative=10,
    noise=0.5,
    random_state=42,
)
```

```python
rf_reg_space = SearchSpace(
    name="rf_regressor_optuna",
    parameters={
        "n_estimators": [
            50,
            100,
            200,
        ],
        "max_depth": [
            3,
            5,
            10,
            None,
        ],
        "min_samples_split": [
            2,
            5,
            10,
        ],
    },
)
```

```python
result = optimizer.optimize(
    algorithm="random_forest_regressor",
    X=X,
    y=y,
    metric="mse",
    cv=5,
    n_trials=20,
    search_space=rf_reg_space,
)
```

```python
print(
    f"Best MSE: {result.display_score:.4f}"
)

print(
    result.best_params
)
```

```python
best_model = result.best_model

predictions = best_model.predict(
    X[:10]
)

predictions
```
