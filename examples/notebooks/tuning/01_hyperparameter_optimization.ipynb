```python
from sklearn.datasets import make_classification

from mlcore.tuning.sklearn.grid import (
    GridSearchOptimizer,
)

from mlcore import MODEL_REGISTRY
```

```python
X, y = make_classification(
    n_samples=1000,
    n_features=20,
    n_informative=10,
    random_state=42,
)
```

```python
optimizer = GridSearchOptimizer(
    registry=MODEL_REGISTRY,
)
```

```python
result = optimizer.optimize(
    algorithm="random_forest",
    X=X,
    y=y,
    metric="f1",
    cv=5,
)
```

```python
result.best_score
```

```python
result.best_params
```

```python
result.summary()
```

```python
import pandas as pd

pd.DataFrame(
    result.history
).sort_values(
    "mean_test_score",
    ascending=False,
).head()
```

```python
from mlcore.core.search_space import (
    SearchSpace,
)
```

```python
custom_space = SearchSpace(
    name="rf_custom",
    parameters={
        "n_estimators": [50, 100],
        "max_depth": [3, 5],
    },
)
```

```python
X, y = make_classification(
    n_samples=1000,
    n_features=20,
    n_informative=10,
    random_state=42,
)
```

```python
result = optimizer.optimize(
    algorithm="random_forest",
    X=X,
    y=y,
    metric="accuracy",
    cv=5,
    search_space=custom_space,
)
```

```python
result.best_params
```
